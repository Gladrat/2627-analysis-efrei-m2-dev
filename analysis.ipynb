# Agreger

```python
sessions_per_station = sessions.groupby("station_id")
sessions_per_station.size()
```

```python
sessions_per_station = (
    sessions.groupby("station_id")
    .size()
    .sort_values(ascending=False)
)
sessions_per_station
```

```python
sessions.groupby("station_id")["ended_at"].count()
```

```python
completed = sessions[sessions["status"] == "completed"].copy()
```

```python
station_summary = (
    completed.groupby("station_id", as_index=True)
        .agg(
            session_count=("session_id", "size"),
            total_energy_kwh=("energy_kwh", "sum"),
            mean_energy_kwh=("energy_kwh", "mean"),
        )
).sort_values("total_energy_kwh", ascending=False).round(2)
station_summary
```
